```python
#colocar pontuaÃ§Ã£o portuguÃªsa em stopwordsnltk
import nltk
stopwords = ['a', 'agora', 'algum', 'alguma', 'aquele', 'aqueles', 'de', 'deu', 'do', 'e', 'estou', 'esta', 'esta',
             'ir', 'meu', 'muito', 'mesmo', 'no', 'nossa', 'o', 'outro', 'para', 'que', 'sem', 'talvez', 'tem', 'tendo',
             'tenha', 'teve', 'tive', 'todo', 'um', 'uma', 'umas', 'uns', 'vou']
from nltk.corpus import stopwords
from string import punctuation
nltk.download('stopwords')
stopwordsnltk = set(stopwords.words('portuguese') + list(punctuation))
print(stopwordsnltk)
```

```python
#remoÃ§Ã£o de stopwords
def removestopwords(texto):
    frases = []
    for (palavras, emocao) in texto:
        semstop = [p for p in palavras.split() if p not in stopwordsnltk]
        frases.append((semstop, emocao))
    return frases
```

```python
import nltk
nltk.download('rslp', quiet=True)
#stemmizaÃ§Ã£o
def aplicastemmer(texto):
    stemmer = nltk.stem.RSLPStemmer()
    frasessstemming = []
    for (palavras, emocao) in texto:
        comstemming = [str(stemmer.stem(p)) for p in palavras.split() if p not in stopwordsnltk]
        frasessstemming.append((comstemming, emocao))
    return frasessstemming
#aplicar stemmer e remoÃ§Ã£o de stopwords nas duas bases, teste e treinamento
frasescomstemmtreinamento = aplicastemmer(basetreinamento)
frasescomstemmteste = aplicastemmer(baseparatestes)
```

```python
#criar funÃ§Ã£o para buscar palavras antes da vÃ­rgula para procurar frequencia nas duas bases, teste e treinamento
def buscapalavras(frases):
    todaaspalavras = []
    for (palavras, emocao) in frases:
        todaaspalavras.extend(palavras)
    return todaaspalavras
palavrastreinamento = buscapalavras(frasescomstemmtreinamento)
palavrasteste = buscapalavras(frasescomstemmteste)
```

```python
#criar funÃ§Ã£o para buscar frequÃªncia antes da vÃ­rgula para procurar frequencia nas duas bases, teste e treinamento
def buscafrequencia(palavras):
    palavras = nltk.FreqDist(palavras)
    return palavras
frequenciatreinamento = buscafrequencia(palavrastreinamento)
frequenciateste = buscafrequencia(palavrasteste)
print(frequenciatreinamento.most_common(30))
print(frequenciateste.most_common(30))
```

```python
#criar funÃ§Ã£o para buscar palavras Ãºnicas antes da vÃ­rgula para procurar frequencia nas duas bases, teste e treinamento
def procurapalavrasunicas(frequencia):
    freq = frequencia.keys()
    return freq
palavrasunicastreinamento = procurapalavrasunicas(frequenciatreinamento)
palavrasunicasteste = procurapalavrasunicas(frequenciateste)
print(palavrasunicastreinamento)
print(palavrasunicasteste)
```

```python
#criar funÃ§Ã£o para buscar palavras Ãºnicas antes da vÃ­rgula para achar pedaÃ§os na base treinamento
def analisarpalavras(documento):
    doc = set(documento)
    caracteristicas = {}
    for palavras in palavrasunicastreinamento:
        caracteristicas['%s' % palavras] = (palavras in doc)
    return caracteristicas
acharpedacos = analisarpalavras(['admi', 'trem', 'pav'])
print(acharpedacos)
```

```python
#cria classificadores, atravpes do nltk.classify, nas duas bases, teste e treinamento
basecheiatreinamento = nltk.classify.apply_features(analisarpalavras, frasescomstemmtreinamento)
basecheiateste = nltk.classify.apply_features(analisarpalavras, frasescomstemmteste)
```

```python
#treina a BASECHEIATREINAMENTO
classificador = nltk.NaiveBayesClassifier.train(basecheiatreinamento)
print(classificador.labels())

#Testa os dados
from nltk.metrics import ConfusionMatrix
esperado = []
previsto = []
for (frase, classe) in basecheiateste:
    resultado = classificador.classify(frase)
    previsto.append(resultado)
    esperado.append(classe)
matriz = ConfusionMatrix(esperado, previsto)
print(matriz)
```
